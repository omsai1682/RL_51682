```python
import random

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2



N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': 0,
    'B': 0,
    'C': 0
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}


reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}


reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}
total_reward = 0




for play in range(1, N_play + 1):


    r = random.random()


    highest_Q = max(Q.values())


    highest_machines = [
        machine for machine in machines
        if Q[machine] == highest_Q
    ]


    highest_machine = random.choice(highest_machines)


    if r < epsilon:


        other_machines = [
            machine for machine in machines
            if machine != highest_machine
        ]

        selected_machine = random.choice(other_machines)

        decision = "Exploration"


    else:

        selected_machine = highest_machine

        decision = "Exploitation"

    index = reward_index[selected_machine]


    if index >= len(reward_patterns[selected_machine]):
        print("\nAll rewards for machine",
              selected_machine, "have been used.")
        break

    reward = reward_patterns[selected_machine][index]


    reward_index[selected_machine] += 1





    N[selected_machine] += 1
    R[selected_machine] += reward


    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine])
        / N[selected_machine]
    )


    total_reward += reward




    print("\n-----------------------------------------")
    print("Play:", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)
    print("Reward:", reward)

    print("N:", N)
    print("R:", R)
    print("Q:", {
        machine: round(Q[machine], 2)
        for machine in machines
    })




print("\n\n=========================================")
print("FINAL RESULTS")
print("=========================================")

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        machine,
        "\t",
        N[machine],
        "\t",
        R[machine],
        "\t",
        round(Q[machine], 2)
    )

print("\nTotal Reward:", total_reward)


best_machine = max(Q, key=Q.get)
print("Best Machine:", best_machine)
print("Highest Q-value:", round(Q[best_machine], 2))
```

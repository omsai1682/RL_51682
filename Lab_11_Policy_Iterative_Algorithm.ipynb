```python
# Value Iteration in Deterministic 2x2 Grid World

# Part 1: Initialize parameters
gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

# Initial state values
V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

# Part 2: Define transitions
# Format: state -> action -> (next state, reward)

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },

    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

# Part 3: Value Iteration

iteration = 0

print("Iteration-wise State Values")
print("--------------------------------")

while True:

    iteration += 1
    new_V = V.copy()
    delta = 0

    for state in ['S0', 'S1']:

        q_values = []

        for action in actions:

            next_state, reward = transitions[state][action]

            q = reward + gamma * V[next_state]

            q_values.append(q)

        new_V[state] = max(q_values)

        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V

    print(
        f"Iteration {iteration}: "
        f"S0 = {V['S0']:.3f}, "
        f"S1 = {V['S1']:.3f}"
    )

    if delta < theta:
        break

# Part 4: Calculate final Q values

print("\nFinal Action Values")

Q = {}

for state in ['S0', 'S1']:

    Q[state] = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        q = reward + gamma * V[next_state]

        Q[state][action] = q

        print(
            f"{state} - {action}: "
            f"{q:.3f} -> {next_state}"
        )

# Part 5: Find optimal policy

print("\nFinal Optimal State Values")

for state in states:
    print(f"V*({state}) = {V[state]:.3f}")

print("\nOptimal Policy")

for state in ['S0', 'S1']:

    best_action = max(Q[state], key=Q[state].get)

    print(f"{state} -> {best_action}")

print("\nOptimal Path: S0 -> S1 -> G")
```

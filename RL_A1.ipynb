```python
"""
Task 1: Greedy Algorithm for Multi-Armed Bandit
Using the Arithmetic Mean method to estimate Q(a)
"""


def calculate_q_arithmetic_mean(rewards):
    """Q(a) = (Sum of rewards for action a) / (N(a))"""
    return sum(rewards) / len(rewards)


def main():
    n = int(input("Enter the number of machines: "))

    machine_names = []
    machine_rewards = {}

    for i in range(n):
        name = input(f"Enter name of machine {i + 1}: ").strip()
        rewards_str = input(f"Enter observed rewards for machine {name} (comma separated): ")
        rewards = [float(x.strip()) for x in rewards_str.split(",")]
        machine_names.append(name)
        machine_rewards[name] = rewards

    q_values = {}

    print("\n--- Results ---")
    for name in machine_names:
        rewards = machine_rewards[name]
        q = calculate_q_arithmetic_mean(rewards)
        q_values[name] = q
        print(f"\nMachine {name}")
        print(f"Rewards: {rewards}")
        print(f"Q({name}) = {q:.2f}")

    selected_machine = max(q_values, key=q_values.get)
    print(f"\nGreedy Selected Machine: {selected_machine}")


if __name__ == "__main__":
    main()
```

```python
"""
Task 2: Greedy Algorithm for Multi-Armed Bandit
Using the Incremental Update method to estimate Q(a)

Qnew(a) = Qold(a) + (R - Qold(a)) / N(a)
"""


def main():
    n = int(input("Enter the number of machines: "))

    machine_names = []
    machine_rewards = {}

    for i in range(n):
        name = input(f"Enter name of machine {i + 1}: ").strip()
        rewards_str = input(f"Enter observed rewards for machine {name} (comma separated): ")
        rewards = [float(x.strip()) for x in rewards_str.split(",")]
        machine_names.append(name)
        machine_rewards[name] = rewards

    # Task 1: Initialize
    Q = {name: 0.0 for name in machine_names}
    N = {name: 0 for name in machine_names}

    # Task 2: Update after every play, machine by machine
    print("\n--- Step-by-step Updates ---")
    for name in machine_names:
        for reward in machine_rewards[name]:
            N[name] += 1
            Q[name] = Q[name] + (reward - Q[name]) / N[name]
            print(f"Played {name} -> Reward: {reward} | N({name})={N[name]} | Q({name})={Q[name]:.2f}")
            current_best = max(Q, key=Q.get)
            print(f"   Current best machine: {current_best} (Q={Q[current_best]:.2f})")

    print("\n--- Final Estimated Values ---")
    for name in machine_names:
        print(f"Q({name}) = {Q[name]:.2f}")

    selected_machine = max(Q, key=Q.get)
    print(f"\nGreedy Selected Machine: {selected_machine}")


if __name__ == "__main__":
    main()

    # --- Demo run with the assignment's sample data (uncomment to test directly) ---
    # sample_data = {"A": [3, 2, 4], "B": [6, 7], "C": [5, 4, 6]}
    # Q = {name: 0.0 for name in sample_data}
    # N = {name: 0 for name in sample_data}
    # for name, rewards in sample_data.items():
    #     for reward in rewards:
    #         N[name] += 1
    #         Q[name] = Q[name] + (reward - Q[name]) / N[name]
    #         print(f"Played {name} -> Reward: {reward} | N={N[name]} | Q({name})={Q[name]:.2f}")
    # print("\nFinal Q values:", {k: round(v, 2) for k, v in Q.items()})
    # print("Greedy Selected Machine:", max(Q, key=Q.get))
```

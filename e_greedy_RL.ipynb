```python
import random

machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

plays = int(input("Enter number of plays: "))

epsilon = float(input("Enter epsilon (0 to 1): "))

N = {"A": 0, "B": 0, "C": 0}
Q = {"A": 0.0, "B": 0.0, "C": 0.0}

print("\nPlay\tMachine\tReward\tEstimated Q\tAction")

for i in range(1, plays + 1):

    r = random.random()

    if r < epsilon:
        machine = random.choice(list(machines.keys()))
        action = "Explore"

    else:
        machine = max(Q, key=Q.get)
        action = "Exploit"

    # Get reward
    reward = random.choice(machines[machine])

    N[machine] += 1

    Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

    print(f"{i}\t{machine}\t{reward}\t{Q[machine]:.2f}\t\t{action}")

print("\nFinal Results")
print("----------------------------")
for machine in machines:
    print(f"Machine {machine}: Selected = {N[machine]}, Estimated Reward = {Q[machine]:.2f}")
```
